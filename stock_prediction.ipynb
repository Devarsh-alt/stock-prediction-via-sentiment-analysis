**PULLING NASDAQ 100 STOCKS**


This code scrapes the ticker symbols of NASDAQ-100 companies from Wikipedia. It starts by storing the URL of the NASDAQ-100 Wikipedia page in the variable url, then uses pd.read_html(url) to read all HTML tables on the page into a list of DataFrames. The [4] index accesses the fifth table (which contains the list of companies), ['Ticker'] selects the "Ticker" column from that table, and .tolist() converts it into a standard Python list. Finally, print(stockslist) displays the list of ticker symbols.

```python

url = "https://en.wikipedia.org/wiki/NASDAQ-100"
stockslist = pd.read_html(url)[4]['Ticker'].tolist()

print(stockslist)
```

```python
len(stockslist)
```

This code creates a dictionary hists to store the 3-year historical stock data for each company in the stockslist. It loops through each ticker symbol s in the list, uses yf.Ticker(s) to create a Ticker object from the yfinance library, then calls .history(period="3y") to fetch the past 3 years of historical stock data (including Open, High, Low, Close, Volume, etc.). The resulting DataFrame is stored in the dictionary hists with the ticker symbol s as the key.

```python
hists = {}
for s in stockslist:
    tkr = yf.Ticker(s)
    history = tkr.history(period="3y")
    hists[s] = history
```

This code generates and displays interactive candlestick charts for each stock in the stockslist using Plotly. It loops through each ticker symbol, copies its historical data from the hists dictionary into temp_df, and then creates a candlestick chart using go.Figure with go.Candlestick, mapping the date (x) to the index and stock price data to the open, high, low, and close values. The layout is customized with margins, height, background color, and a title set to the stock's ticker symbol. Finally, fig.show() renders the chart interactively for each stock, one at a time.

```python
for stock in stockslist:
    temp_df = hists[stock].copy()

    fig = go.Figure(
        data=[
            go.Candlestick(
                x=temp_df.index,
                open=temp_df["Open"],
                high=temp_df["High"],
                low=temp_df["Low"],
                close=temp_df["Close"],
            )
        ]
    )

    fig.update_layout(
        margin=dict(l=20, r=20, t=60, b=20),
        height=300,
        paper_bgcolor="LightSteelBlue",
        title=stock,
    )

    fig.show()
```

**USER CHOICE GRAPH DISPLAY**


This Python script creates an interactive stock price visualization tool using matplotlib, mplfinance, and ipywidgets within a Jupyter Notebook. It allows users to select any stock from a list (stockslist) using a dropdown menu and choose from three chart types—Line Chart, OHLC Chart, and Area Chart—using buttons. The hists dictionary stores historical stock data, which is used to generate visualizations based on user interaction. When a chart type is selected, the corresponding function (plot_line_chart, plot_ohlc_chart, or plot_area_chart) is triggered, which clears previous output and displays a new chart for the selected stock. Line and area charts are created using matplotlib, while OHLC charts use the mplfinance library. The UI is built with ipywidgets, and dynamic updates are handled using clear_output and display. A default line chart is also shown for the first stock when the interface loads.




When the code is executed, it first imports all necessary libraries and defines the list of stocks from the hists dictionary. It then initializes a dropdown widget and three buttons for selecting the chart type. An empty output area is also created to display the charts. After setup, the line chart for the first stock in the list is plotted by calling plot_line_chart(stockslist[0]), which immediately shows the default output. Then, the button click event handlers are registered: clicking a button triggers its respective function (plot_line_chart, plot_ohlc_chart, or plot_area_chart), which fetches the selected stock from the dropdown, copies its data, clears the output, and displays the selected chart. The display() function renders the dropdown, buttons, and output area in the notebook, and from that point onward, any interaction by the user dynamically updates the chart area based on the selected stock and chart type.



```python
import matplotlib.pyplot as plt
import mplfinance as mpf
import ipywidgets as widgets
from IPython.display import display, clear_output

# Sample stocks list (replace with your actual list)
stockslist = list(hists.keys())

# Dropdown for selecting stock
dropdown = widgets.Dropdown(
    options=stockslist,
    description="Select Stock:",
    style={"description_width": "initial"},
)

# Buttons for selecting chart type
btn_line = widgets.Button(description="Line Chart")
btn_ohlc = widgets.Button(description="OHLC Chart")
btn_area = widgets.Button(description="Area Chart")

# Output area for displaying the chart
output = widgets.Output()

def plot_line_chart(stock):
    temp_df = hists[stock].copy()

    with output:
        clear_output(wait=True)
        plt.figure(figsize=(10, 5))
        plt.plot(temp_df.index, temp_df["Close"], label="Close Price", color="blue")
        plt.xlabel("Date")
        plt.ylabel("Price")
        plt.title(f"{stock} Closing Price")
        plt.legend()
        plt.grid(True)
        plt.show()

def plot_ohlc_chart(stock):
    temp_df = hists[stock].copy()

    with output:
        clear_output(wait=True)
        mpf.plot(
            temp_df,
            type="ohlc",
            style="charles",
            title=f"{stock} OHLC Chart",
            volume=False
        )

def plot_area_chart(stock):
    temp_df = hists[stock].copy()

    with output:
        clear_output(wait=True)
        plt.figure(figsize=(10, 5))
        plt.fill_between(temp_df.index, temp_df["Close"], color="skyblue", alpha=0.4)
        plt.plot(temp_df.index, temp_df["Close"], color="blue", alpha=0.6, linewidth=2)
        plt.xlabel("Date")
        plt.ylabel("Price")
        plt.title(f"{stock} Price Movement")
        plt.grid(True)
        plt.show()

# Button click event handlers
def on_line_chart_click(b):
    plot_line_chart(dropdown.value)

def on_ohlc_chart_click(b):
    plot_ohlc_chart(dropdown.value)

def on_area_chart_click(b):
    plot_area_chart(dropdown.value)

# Attach button events
btn_line.on_click(on_line_chart_click)
btn_ohlc.on_click(on_ohlc_chart_click)
btn_area.on_click(on_area_chart_click)

# Display UI elements
display(dropdown, btn_line, btn_ohlc, btn_area, output)

# Default plot on start
plot_line_chart(stockslist[0])
```

**IMPORTING NEWS ARTICLE SCRAAPING NEWS VIA GOOGLE NEWS FOR EACH STOCK**


This code is designed to scrape recent news articles related to specific stocks from Google News and store them for further sentiment analysis or display. It begins by defining a function get_news() that takes a stock name and queries Google News using requests, with browser-like headers to avoid being blocked. The HTML content is parsed using BeautifulSoup, and multiple robust CSS selectors are used to extract news items (title, link, source, publish time). The function ensures only valid and clean URLs are collected, handling Google redirect links when needed. It limits the number of articles to max_articles (default 5). After defining the function, the script loops through each stock symbol in stockslist2, scrapes the news using get_news(), and stores the results in the all_news dictionary. To prevent IP bans, it introduces a random delay between requests using time.sleep() with a delay between 2 to 4 seconds. Then, it converts the scraped articles for each stock into a pandas DataFrame and stores them in a dictionary called news_data for easier manipulation and analysis. Finally, it displays a sample of two news articles for the first three stocks as a preview. This code provides both raw data (as a list of dictionaries in all_news) and structured tabular format (in news_data) for news related to stocks like AAPL, MSFT, etc., allowing easy access for further use in sentiment analysis or visualization

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from newspaper import Article
import time
import random

# Function to scrape Google News for a stock with more robust CSS selectors
def get_news(stock, max_articles=5):
    query = f"{stock} stock news"
    url = f"https://www.google.com/search?q={query}&tbm=nws"
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36",
        "Accept-Language": "en-US,en;q=0.9",
        "Referer": "https://www.google.com/",
        "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8"
    }

    articles = []
    try:
        response = requests.get(url, headers=headers)
        response.raise_for_status()

        soup = BeautifulSoup(response.text, "html.parser")

        # Try multiple potential selectors for news items
        news_items = (
            soup.select("div.SoaBEf") or
            soup.select("div.iRPxbe") or
            soup.select("div.WlydOe") or
            soup.select("div[role='heading']") or
            soup.select("g-card")
        )

        for item in news_items:
            if len(articles) >= max_articles:
                break

            # Try different approaches to extract title and URL
            title_element = (
                item.select_one("div.n0jPhd") or
                item.select_one("div.mCBkyc") or
                item.select_one("h3") or
                item.select_one("a[data-ved]")
            )

            link_element = item.select_one("a")
            source_element = (
                item.select_one("div.MgUUmf") or
                item.select_one("span.CEMjEf") or
                item.select_one("div.CEMjEf")
            )

            time_element = (
                item.select_one("div.OSrXXb") or
                item.select_one("span.WG9SHc") or
                item.select_one("span.LfVVr")
            )

            if title_element and link_element and hasattr(link_element, 'get'):
                title = title_element.get_text(strip=True)
                link = link_element.get("href", "")
                source = source_element.get_text(strip=True) if source_element else "Unknown"
                publish_time = time_element.get_text(strip=True) if time_element else "Unknown"

                # Clean the URL - Google News URLs may contain redirects
                if link.startswith("/url?"):
                    from urllib.parse import parse_qs
                    link = parse_qs(link.split("?", 1)[1]).get('q', [link])[0]

                if title and link:
                    articles.append({
                        "Title": title,
                        "Source": source,
                        "Published": publish_time,
                        "URL": link
                    })

    except Exception as e:
        print(f"Error scraping {stock}: {e}")

    return articles

# Dictionary to store news data
all_news = {}

# Scrape news for each stock in stockslist2
for stock in stockslist2:
    print(f"Scraping news for {stock}...")
    stock_news = get_news(stock)

    if stock_news:
        all_news[stock] = stock_news
        print(f"  Found {len(stock_news)} articles")
    else:
        print(f"  No articles found")

    # Random delay between requests to avoid being blocked
    delay = random.uniform(2, 4)
    time.sleep(delay)

# Create DataFrames for each stock (stored in a dictionary)
news_data = {}
for stock, articles in all_news.items():
    if articles:
        news_data[stock] = pd.DataFrame(articles)

# Display sample results
for stock in list(all_news.keys())[:3]:
    if stock in news_data:
        print(f"\nSample News for {stock}:")
        print(news_data[stock].head(2))
    else:
        print(f"\nNo news found for {stock}")

# Now you can access the news for any stock like this:
# news_data["AAPL"]  # Returns DataFrame with Apple news
# all_news["MSFT"]   # Returns list of dictionaries with Microsoft news
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from newspaper import Article
import time
import random
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer
import yfinance as yf
from datetime import datetime, timedelta
import matplotlib.pyplot as plt

# Download NLTK resources
nltk.download('vader_lexicon')

# Initialize sentiment analyzer
sia = SentimentIntensityAnalyzer()

# Sample stock list (replace with your actual list)
stockslist2 = ["AAPL", "MSFT", "GOOGL", "AMZN", "META"]
```

The `get_news()` function scrapes the latest news headlines for a given stock from Google News. It starts by forming a search query using the stock name and constructs a Google News URL. To mimic a browser and avoid being blocked, it sends the request with custom headers using the `requests` library. Once the response is received, it parses the HTML using `BeautifulSoup`. Since Google News may change its layout frequently, the function uses multiple CSS selectors to robustly locate relevant news items. For each news block, it attempts to extract the title, article link, source, and published time using a combination of alternative HTML tags and class names. It also handles cleaning of URLs that start with `/url?`, which are often redirect links, by extracting the actual URL from the query string. The function collects this information into a dictionary for each article and appends it to the `articles` list, stopping once the specified maximum number of articles (`max_articles`) is reached. If an error occurs during the process, it's caught and printed. Finally, the function returns a list of dictionaries, each representing a news article related to the stock.

```python
def get_news(stock, max_articles=5):
    query = f"{stock} stock news"
    url = f"https://www.google.com/search?q={query}&tbm=nws"
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36",
        "Accept-Language": "en-US,en;q=0.9",
        "Referer": "https://www.google.com/",
        "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8"
    }

    articles = []
    try:
        response = requests.get(url, headers=headers)
        response.raise_for_status()

        soup = BeautifulSoup(response.text, "html.parser")

        # Try multiple potential selectors for news items
        news_items = (
            soup.select("div.SoaBEf") or
            soup.select("div.iRPxbe") or
            soup.select("div.WlydOe") or
            soup.select("div[role='heading']") or
            soup.select("g-card")
        )

        for item in news_items:
            if len(articles) >= max_articles:
                break

            # Try different approaches to extract title and URL
            title_element = (
                item.select_one("div.n0jPhd") or
                item.select_one("div.mCBkyc") or
                item.select_one("h3") or
                item.select_one("a[data-ved]")
            )

            link_element = item.select_one("a")
            source_element = (
                item.select_one("div.MgUUmf") or
                item.select_one("span.CEMjEf") or
                item.select_one("div.CEMjEf")
            )

            time_element = (
                item.select_one("div.OSrXXb") or
                item.select_one("span.WG9SHc") or
                item.select_one("span.LfVVr")
            )

            if title_element and link_element and hasattr(link_element, 'get'):
                title = title_element.get_text(strip=True)
                link = link_element.get("href", "")
                source = source_element.get_text(strip=True) if source_element else "Unknown"
                publish_time = time_element.get_text(strip=True) if time_element else "Unknown"

                # Clean the URL - Google News URLs may contain redirects
                if link.startswith("/url?"):
                    from urllib.parse import parse_qs
                    link = parse_qs(link.split("?", 1)[1]).get('q', [link])[0]

                if title and link:
                    articles.append({
                        "Title": title,
                        "Source": source,
                        "Published": publish_time,
                        "URL": link
                    })

    except Exception as e:
        print(f"Error scraping {stock}: {e}")

    return articles
```

```python
# Dictionary to store news data
all_news = {}

# Scrape news for each stock in stockslist2
for stock in stockslist2:
    print(f"Scraping news for {stock}...")
    stock_news = get_news(stock)

    if stock_news:
        all_news[stock] = stock_news
        print(f"  Found {len(stock_news)} articles")
    else:
        print(f"  No articles found")

    # Random delay between requests to avoid being blocked
    delay = random.uniform(2, 4)
    time.sleep(delay)

# Create DataFrames for each stock (stored in a dictionary)
news_data = {}
for stock, articles in all_news.items():
    if articles:
        news_data[stock] = pd.DataFrame(articles)

# Display sample results
for stock in list(all_news.keys())[:3]:
    if stock in news_data:
        print(f"\nSample News for {stock}:")
        print(news_data[stock].head(2))
    else:
        print(f"\nNo news found for {stock}")
```

This code first defines two helper functions: extract_article_text(url) uses the newspaper3k library to download and parse a news article from a given URL, returning its plain text content, while get_stock_data(ticker, period) uses the yfinance library to download recent stock price data (defaulting to the past 1 month). The script then initializes an empty dictionary, stock_data_dict, and populates it by fetching stock data for each symbol in stockslist2. If the data is successfully retrieved, it is stored in the dictionary; otherwise, an error message is printed. After stock data is collected, the code proceeds to perform sentiment analysis on the news articles for each stock in the all_news dictionary. For every article, it extracts the full text using the previously defined function, analyzes its sentiment using VADER (sia.polarity_scores()), and stores the resulting sentiment scores—especially the compound score—within the article dictionary. It also prints out the article title (truncated) along with the compound sentiment score to give a real-time summary of the analysis process. This setup enables a direct mapping between sentiment data and corresponding stock tickers for further prediction and visualization.

```python
def extract_article_text(url):
    try:
        article = Article(url)
        article.download()
        article.parse()
        return article.text
    except Exception as e:
        print(f"Error extracting article text: {e}")
        return ""

def get_stock_data(ticker, period="1mo"):
    try:
        stock_data = yf.download(ticker, period=period)
        return stock_data
    except Exception as e:
        print(f"Error fetching stock data for {ticker}: {e}")
        return None

# Get stock data for all stocks
stock_data_dict = {}
for stock in stockslist2:
    stock_data = get_stock_data(stock)
    if stock_data is not None:
        stock_data_dict[stock] = stock_data
        print(f"Got data for {stock} ({len(stock_data)} days)")
    else:
        print(f"Failed to get data for {stock}")

# Analyze sentiment for all articles
for stock in all_news:
    print(f"\nAnalyzing sentiment for {stock}...")
    for article in all_news[stock]:
        article_text = extract_article_text(article['URL'])
        sentiment = sia.polarity_scores(article_text)
        article['Sentiment'] = sentiment
        print(f"  {article['Title'][:50]}...: {sentiment['compound']:.2f}")
```

```python
# Plot sentiment vs stock price for each stock
for stock in stockslist2:
    if stock in stock_data_dict and stock in all_news:
        plt.figure(figsize=(12, 5))

        # Plot stock price
        plt.subplot(1, 2, 1)
        stock_data_dict[stock]['Close'].plot(title=f"{stock} Price")

        # Plot sentiment scores
        plt.subplot(1, 2, 2)
        sentiments = [a['Sentiment']['compound'] for a in all_news[stock]]
        dates = [datetime.now() - timedelta(days=i) for i in range(len(sentiments))][::-1]
        plt.plot(dates, sentiments, 'o-')
        plt.title(f"{stock} News Sentiment")
        plt.xticks(rotation=45)

        plt.tight_layout()
        plt.show()
```

This code generates a summary analysis of sentiment and stock price performance for multiple companies by iterating through a list of selected stock tickers (stockslist2). For each stock, it first ensures that both sentiment data (all_news) and historical price data (stock_data_dict) are available and valid. If any data is missing or insufficient (like no articles or less than two price points), it skips that stock with a message. For valid entries, it computes the average sentiment score using the compound values from VADER sentiment analysis, and calculates the percentage price change between the first and last closing price in the stock’s history. These results are stored in a summary_data list. Once all stocks are processed, the list is converted into a pandas DataFrame called summary_df. If there is enough data, the summary is printed and sorted by sentiment score. Additionally, if there are results for at least two stocks, a scatter plot is generated showing sentiment vs. price change, with each point labeled by the stock symbol to visualize any potential correlation between news sentiment and stock performance. If no usable data is found, the script outputs a message saying so.

```python
# Create summary DataFrame with robust error handling
summary_data = []
for stock in stockslist2:
    try:
        # Check if we have both news and price data
        if stock not in all_news or stock not in stock_data_dict:
            print(f"Skipping {stock} - missing data")
            continue

        # Check if we have articles
        if len(all_news[stock]) == 0:
            print(f"Skipping {stock} - no articles found")
            continue

        # Check if we have price data
        if len(stock_data_dict[stock]) == 0:
            print(f"Skipping {stock} - no price data")
            continue

        # Calculate average sentiment
        avg_sentiment = sum([a.get('Sentiment', {}).get('compound', 0)
                          for a in all_news[stock]]) / len(all_news[stock])

        # Calculate price change with index validation
        price_data = stock_data_dict[stock]['Close']
        if len(price_data) < 2:
            print(f"Skipping {stock} - insufficient price history")
            continue

        price_change = (price_data.iloc[-1] - price_data.iloc[0]) / price_data.iloc[0] * 100

        summary_data.append({
            'Stock': stock,
            'Articles': len(all_news[stock]),
            'Avg Sentiment': avg_sentiment,
            'Price Change (%)': price_change
        })

    except Exception as e:
        print(f"Error processing {stock}: {str(e)[:100]}...")
        continue

# Create and display DataFrame if we have data
if summary_data:
    summary_df = pd.DataFrame(summary_data)
    print("\nSummary Report:")
    display(summary_df.sort_values('Avg Sentiment', ascending=False))

    # Correlation plot if we have at least 2 stocks
    if len(summary_df) >= 2:
        plt.figure(figsize=(8, 5))
        plt.scatter(summary_df['Avg Sentiment'], summary_df['Price Change (%)'])

        # Add stock labels
        for i, row in summary_df.iterrows():
            plt.text(row['Avg Sentiment']+0.01,
                    row['Price Change (%)']+0.5,
                    row['Stock'],
                    fontsize=9)

        plt.title("Sentiment vs Price Change")
        plt.xlabel("Average Sentiment Score")
        plt.ylabel("Price Change (%)")
        plt.grid(True)
        plt.show()
    else:
        print("Not enough data for correlation plot")
else:
    print("No valid data to create summary report")
```

This code defines a simple prediction model that estimates stock price movement over the next three days based on the average sentiment score of news articles. The function predict_next_days() takes a summary DataFrame (which contains stocks, their average sentiment, and historical price change) and a number of future days to predict (default is 3). It initializes a predictions DataFrame with the stock names and their sentiment scores, and sets the starting value (Day 0) to 0% change. Using a loop, it then generates predicted daily changes by doubling the sentiment score (acting as a base growth factor), and introduces randomness via a noise factor that varies ±30% to simulate market unpredictability. These daily predictions are accumulated to reflect cumulative percentage changes over time. The function returns the prediction DataFrame. The code then generates predictions using this model and visualizes them for up to five stocks. It plots each stock’s predicted price change over 3 days using matplotlib, showing lines labeled with the stock ticker and its sentiment score. This simple model illustrates how positive sentiment could translate into expected short-term gains, with built-in variability mimicking real-world market behavior.

```python
# Simple prediction model using sentiment scores with natural variability
def predict_next_days(summary_df, days=3):
    import numpy as np

    # Create prediction results DataFrame
    predictions = pd.DataFrame()
    predictions['Stock'] = summary_df['Stock']
    predictions['Current Sentiment'] = summary_df['Avg Sentiment'].round(4)

    # Set random seed for reproducibility
    np.random.seed(42)

    # Starting point is 0 (today)
    predictions['Day 0'] = 0

    # Calculate simple prediction based on sentiment correlation
    cumulative_change = np.zeros(len(summary_df))

    for day in range(1, days+1):
        # Base prediction: Higher sentiment = higher predicted growth
        base_prediction = summary_df['Avg Sentiment'] * 2

        # Add random variability (±30% of the base prediction)
        noise_factor = np.random.uniform(0.7, 1.3, len(summary_df))
        daily_change = base_prediction * noise_factor

        # Accumulate changes (cumulative effect)
        cumulative_change += daily_change

        # Store cumulative prediction
        predictions[f'Day {day} Pred (%)'] = cumulative_change

    return predictions

# Generate predictions
prediction_df = predict_next_days(summary_df)

# Display predictions
print("\nStock Price Predictions Based on Sentiment (Next 3 Days):")
display(prediction_df)

# Visualize predictions for all stocks or max 5 if there are many
stocks_to_plot = min(5, len(prediction_df))
top_stocks = prediction_df.head(stocks_to_plot)
plt.figure(figsize=(10, 6))

for i, row in top_stocks.iterrows():
    stock = row['Stock']
    sentiment = row['Current Sentiment']
    values = [row['Day 0']] + [row[f'Day {day} Pred (%)'] for day in range(1, 4)]
    plt.plot(range(4), values, marker='o', label=f"{stock} (Sentiment: {sentiment:.2f})")

plt.title("3-Day Price Change Predictions")
plt.xlabel("Days")
plt.ylabel("Predicted Cumulative Price Change (%)")
plt.xticks(range(4), ['Today'] + [f'Day {i}' for i in range(1, 4)])
plt.grid(True)
plt.legend()
plt.show()
```
